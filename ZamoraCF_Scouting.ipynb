### 🔧 Mejoras en la carga del Excel

El código original funciona, pero vamos a hacerlo más **robusto y reutilizable**:
- Ruta flexible (no hardcodeada)
- Manejo de errores
- Función de conversión al formato JSON de la app

````python
import os

# ── CONFIGURACIÓN ─────────────────────────────────────────────────────────
# Cambia solo esta variable con la ruta a tu carpeta de Excel
RUTA_EXCEL = os.path.expanduser("~/Downloads/DATABASE WYSCOUT")

# Archivos disponibles
ARCHIVOS_EXCEL = {
    "1ª RFEF":         "1 RFEF - WYSCOUT.xlsx",
    "Hypermotion":     "HYPERMOTION - WYSCOUT.xlsx",
    "LaLiga":          "LALIGA - WYSCOUT.xlsx",
    "5 Grandes Ligas": "5 GRANDES LIGAS - WYSCOUT.xlsx",
}

def cargar_excel(liga, ruta_base=RUTA_EXCEL):
    """
    Carga un Excel de Wyscout de forma segura.
    Devuelve un DataFrame o None si el archivo no existe.
    """
    nombre = ARCHIVOS_EXCEL.get(liga)
    if not nombre:
        print(f"✗ Liga '{liga}' no reconocida. Opciones: {list(ARCHIVOS_EXCEL.keys())}")
        return None
    
    ruta = os.path.join(ruta_base, nombre)
    
    if not os.path.exists(ruta):
        print(f"✗ Archivo no encontrado: {ruta}")
        print(f"  Verifica que RUTA_EXCEL apunta a la carpeta correcta")
        return None
    
    df = pd.read_excel(ruta)
    print(f"✓ {liga} cargado: {len(df)} jugadores · {df.shape[1]} columnas")
    return df

# ── PRUEBA ────────────────────────────────────────────────────────────────
df = cargar_excel("1ª RFEF")
if df is not None:
    print(f"\nColumnas: {list(df.columns)}")
    display(df.head(3))
````

### 🔄 Convertir Excel → JSON de la app

Esta función convierte el DataFrame de Wyscout al formato que usa la app:

````python
# Mapeo de posiciones Wyscout → posiciones de la app
POS_MAP = {
    "GK":   "Portero",
    "RB":   "Lateral Derecho",  "RWB": "Lateral Derecho",
    "LB":   "Lateral Izquierdo","LWB": "Lateral Izquierdo",
    "CB":   "Central", "RCB": "Central", "LCB": "Central",
    "DMF":  "Mediocentro Defensivo","RDMF":"Mediocentro Defensivo","LDMF":"Mediocentro Defensivo",
    "RCMF": "Mediocentro","LCMF":"Mediocentro","CMF":"Mediocentro",
    "AMF":  "Mediapunta",
    "RW":   "Extremo Derecho","RAMF":"Extremo Derecho","RWF":"Extremo Derecho",
    "LW":   "Extremo Izquierdo","LAMF":"Extremo Izquierdo","LWF":"Extremo Izquierdo",
    "CF":   "Delantero Centro","SS":"Segunda Punta",
}

def mapear_posicion(raw):
    """Convierte el código de posición de Wyscout a la posición de la app"""
    if not raw or str(raw) == "nan": return ""
    codigos = [c.strip() for c in str(raw).split(",")]
    for c in codigos:
        if c in POS_MAP:
            return POS_MAP[c]
    return codigos[0] if codigos else ""

def formatear_valor_mercado(v):
    """Convierte valor numérico a texto legible (ej: 5000000 → '5M €')"""
    try:
        v = int(v)
        if v >= 1_000_000: return f"{v//1_000_000}M €"
        if v >= 1_000:     return f"{v//1_000}K €"
        return f"{v} €"
    except:
        return ""

def formatear_pie(f):
    """Normaliza el pie dominante"""
    if not f or str(f) == "nan": return ""
    f = str(f).lower()
    if any(x in f for x in ["der","right"]): return "Derecho"
    if any(x in f for x in ["izq","left"]):  return "Izquierdo"
    if any(x in f for x in ["amb","both"]):  return "Ambidiestro"
    return f.capitalize()

def df_a_jugadores(df, liga="1ª RFEF"):
    """
    Convierte un DataFrame de Wyscout al formato JSON de la app.
    
    Parámetros:
        df   : DataFrame cargado desde el Excel de Wyscout
        liga : Nombre de la liga (ej: '1ª RFEF', 'LaLiga')
    
    Devuelve:
        Lista de diccionarios en el formato de la app
    """
    import time
    jugadores = []
    
    for i, row in df.iterrows():
        # Nombre del jugador (busca columnas posibles)
        nombre = row.get("Jugador", row.get("Player", row.get("Name", "")))
        if not nombre or str(nombre) == "nan":
            continue
        
        # Contrato (puede ser NaN)
        contrato = row.get("Vencimiento contrato", row.get("Contract expires", ""))
        if str(contrato) == "nan": contrato = ""
        
        # Valor de mercado
        valor_raw = row.get("Valor de mercado (Transfermarkt)", row.get("Market value", 0))
        
        j = {
            "id":           f"xls_{liga[:3].lower()}_{i}_{int(time.time())}",
            "name":         str(nombre).strip(),
            "club":         str(row.get("Equipo", row.get("Team", ""))).replace("nan","").strip(),
            "position":     mapear_posicion(row.get("Posición específica", row.get("Position",""))),
            "positionRaw":  str(row.get("Posición específica", row.get("Position",""))).replace("nan",""),
            "age":          int(row.get("Edad", row.get("Age", 0)) or 0),
            "nationality":  str(row.get("País de nacimiento", row.get("Nationality",""))).split(",")[0].strip().replace("nan",""),
            "foot":         formatear_pie(row.get("Pie", row.get("Foot",""))),
            "height":       int(row.get("Altura", row.get("Height", 0)) or 0),
            "weight":       int(row.get("Peso", row.get("Weight", 0)) or 0),
            "contractEnd":  str(contrato).strip(),
            "marketValue":  formatear_valor_mercado(valor_raw),
            "onLoan":       str(row.get("En prestamo", row.get("On loan","no"))).lower() in ["sí","si","yes","true"],
            "liga":         liga,
            "source":       "wyscout",
            "scoutRating":  0,
            "notes":        "",
            "videoLinks":   "",
            "addedBy":      "python",
            "addedAt":      "2025-01-01",
            "stats": {
                "goals":   int(row.get("Goles", row.get("Goals", 0)) or 0),
                "assists": int(row.get("Asistencias", row.get("Assists", 0)) or 0),
                "matches": int(row.get("Partidos jugados", row.get("Matches played", 0)) or 0),
                "minutes": int(row.get("Minutos jugados", row.get("Minutes played", 0)) or 0),
                "xg":      float(row.get("xG", 0) or 0),
                "rating":  float(row.get("Nota media", row.get("Average rating", 0)) or 0),
            }
        }
        jugadores.append(j)
    
    print(f"✓ Convertidos {len(jugadores)} jugadores de '{liga}'")
    return jugadores

print("✅ Funciones de conversión cargadas")
````

## 🔍 Funciones de Consulta

````python
# ── FUNCIONES DE CONSULTA ────────────────────────────────────────────────

def buscar_jugador(jugadores, nombre):
    """Busca jugadores por nombre (parcial, sin distinguir mayúsculas)"""
    return [j for j in jugadores if nombre.lower() in j["name"].lower()]

def filtrar_por_liga(jugadores, liga):
    """Filtra jugadores por liga"""
    return [j for j in jugadores if j.get("liga") == liga]

def filtrar_por_posicion(jugadores, posicion):
    """Filtra jugadores por posición"""
    return [j for j in jugadores if j.get("position") == posicion]

def filtrar_por_edad(jugadores, edad_min, edad_max):
    """Filtra jugadores por rango de edad"""
    return [j for j in jugadores if edad_min <= j.get("age", 0) <= edad_max]

def top_por_valoracion(jugadores, n=10):
    """Devuelve los N jugadores con mayor valoración scout"""
    con_rating = [j for j in jugadores if j.get("scoutRating", 0) > 0]
    return sorted(con_rating, key=lambda j: j["scoutRating"], reverse=True)[:n]

def estadisticas_liga(jugadores):
    """Muestra cuántos jugadores hay por liga"""
    return dict(Counter(j.get("liga", "Sin liga") for j in jugadores))

def contratos_proximos(jugadores, dias=180):
    """Jugadores con contrato que vence en menos de X días"""
    limite = datetime.now() + timedelta(days=dias)
    resultado = []
    for j in jugadores:
        contrato = j.get("contractEnd", "")
        if contrato:
            try:
                fecha = datetime.strptime(contrato, "%Y-%m-%d")
                if fecha <= limite:
                    resultado.append({**j, "dias_restantes": (fecha - datetime.now()).days})
            except: pass
    return sorted(resultado, key=lambda x: x["dias_restantes"])

print("Funciones de consulta cargadas")
````

## ✏️ Funciones de Modificación

````python
# ── FUNCIONES DE MODIFICACIÓN ────────────────────────────────────────────

def añadir_jugador(jugadores, nuevo):
    """Añade un jugador nuevo a la cartera"""
    nuevo.setdefault("id", f"p{int(time.time())}")
    nuevo.setdefault("source", "manual")
    nuevo.setdefault("addedBy", "python")
    nuevo.setdefault("addedAt", datetime.now().strftime("%Y-%m-%d"))
    nuevo.setdefault("videoLinks", "")
    nuevo.setdefault("height", 0)
    nuevo.setdefault("weight", 0)
    nuevo.setdefault("onLoan", False)
    nuevo.setdefault("positionRaw", "")
    nuevo.setdefault("marketValue", "")
    nuevo.setdefault("stats", {"goals":0,"assists":0,"matches":0,"minutes":0,"xg":0.0,"rating":0.0})
    jugadores.append(nuevo)
    print(f"✓ Jugador añadido: {nuevo['name']}")
    return jugadores

def actualizar_valoracion(jugadores, nombre, nueva_valoracion):
    """Actualiza la valoración scout (1-10) de un jugador"""
    for j in jugadores:
        if j["name"].lower() == nombre.lower():
            j["scoutRating"] = nueva_valoracion
            print(f"✓ {j['name']} → valoración actualizada a {nueva_valoracion}/10")
            return jugadores
    print(f"✗ Jugador '{nombre}' no encontrado"); return jugadores

def añadir_nota(jugadores, nombre, nota):
    """Añade o actualiza las notas de un jugador"""
    for j in jugadores:
        if j["name"].lower() == nombre.lower():
            j["notes"] = nota
            print(f"✓ Nota añadida a {j['name']}")
            return jugadores
    print(f"✗ Jugador '{nombre}' no encontrado"); return jugadores

def eliminar_jugador(jugadores, nombre):
    """Elimina un jugador por nombre"""
    antes = len(jugadores)
    jugadores = [j for j in jugadores if j["name"].lower() != nombre.lower()]
    print(f"✓ Jugador '{nombre}' eliminado" if len(jugadores) < antes else f"✗ '{nombre}' no encontrado")
    return jugadores

def añadir_scout(scouts, nombre, password, rol="Scout"):
    """Añade un nuevo scout al sistema"""
    initials = "".join(p[0] for p in nombre.split()[:2]).upper()
    scouts.append({"id":f"s{int(time.time())}","name":nombre,"password":password,"role":rol,"avatar":initials,"favorites":[]})
    print(f"✓ Scout añadido: {nombre} ({rol})")
    return scouts

print("Funciones de modificación cargadas")
````

## 🔄 Regenerar el HTML

Una vez hechos los cambios, usa esta función para actualizar la app:

````python
def regenerar_html(jugadores, scouts, html_entrada="index.html", html_salida="index.html"):
    """
    Inyecta los datos modificados de vuelta en el HTML de la app.
    Genera un nuevo index.html listo para subir a Netlify o GitHub Pages.
    """
    with open(html_entrada, "r", encoding="utf-8") as f:
        html = f.read()

    # Reemplazar DB_PLAYERS
    idx = html.find("const DB_PLAYERS=")
    start = html.find("[", idx)
    depth, end = 0, start
    for i, c in enumerate(html[start:], start):
        if c == '[': depth += 1
        elif c == ']':
            depth -= 1
            if depth == 0: end = i; break
    html = html[:start] + json.dumps(jugadores, ensure_ascii=False) + html[end+1:]

    # Reemplazar DEFAULT_SCOUTS
    idx2 = html.find("const DEFAULT_SCOUTS=")
    start2 = html.find("[", idx2)
    depth, end2 = 0, start2
    for i, c in enumerate(html[start2:], start2):
        if c == '[': depth += 1
        elif c == ']':
            depth -= 1
            if depth == 0: end2 = i; break
    html = html[:start2] + json.dumps(scouts, ensure_ascii=False) + html[end2+1:]

    with open(html_salida, "w", encoding="utf-8") as f:
        f.write(html)
    print(f"HTML regenerado correctamente")
    print(f"   Archivo : {html_salida}")
    print(f"   Tamaño  : {len(html)//1024} KB")
    print(f"   Jugadores: {len(jugadores)}")
    print(f"   Scouts   : {len(scouts)}")
    print(f"\n👉 Sube el nuevo index.html a GitHub o Netlify para publicar los cambios")

print("Función regenerar_html() lista")
````

### 4️⃣ Pizza Chart – Radar de percentiles

````python
# ── PIZZA CHART (Radar de percentiles) ───────────────────────────────────
from mplsoccer import PyPizza

# Métricas y valores del jugador (percentiles 0-100 respecto a su posición)
params = [
    "Goles",  "xG", "Asistencias", "Pases clave",
    "Duelos\nganados", "Presiones", "Recuperaciones",
    "Pases\ncompletados", "Conducción", "Distancia\nrecorrida",
]

# Valores de ejemplo (percentiles)
values_jugador = [82, 75, 68, 71, 55, 60, 58, 80, 73, 65]

# Colores por grupo de métricas
slice_colors = (
    [ZAMORA_RED] * 4 +       # Ataque
    ["#1a3a5c"] * 3 +        # Defensa  
    ["#1a2e1a"] * 3           # Distribución
)
text_colors = [ZAMORA_WHITE] * len(params)

baker = PyPizza(
    params=params,
    background_color=DARK_BG,
    straight_line_color="#3a3a44",
    straight_line_lw=1,
    last_circle_lw=0,
    other_circle_lw=0,
    inner_circle_size=20,
)

fig, ax = baker.make_pizza(
    values_jugador,
    figsize=(9, 9),
    color_blank_space="same",
    slice_colors=slice_colors,
    value_colors=text_colors,
    value_bck_colors=slice_colors,
    blank_alpha=0.15,
    kwargs_slices=dict(edgecolor="#0d0d0f", zorder=2, linewidth=1),
    kwargs_params=dict(color=ZAMORA_WHITE, fontsize=10, va="center"),
    kwargs_values=dict(color=ZAMORA_WHITE, fontsize=9, zorder=3,
                       bbox=dict(edgecolor="#0d0d0f", facecolor=ZAMORA_RED,
                                 boxstyle="round,pad=0.2", lw=1))
)

# Título
fig.text(0.515, 0.975, "Jugador Ejemplo – Delantero Centro",
         size=15, ha="center", color=ZAMORA_WHITE, fontweight='bold')
fig.text(0.515, 0.955, "Percentiles respecto a Delanteros | 1ª RFEF | 2024/25",
         size=10, ha="center", color="#6b6b78")

# Leyenda de grupos
fig.text(0.25, 0.02, "● Ataque", color=ZAMORA_RED, fontsize=10, ha="center")
fig.text(0.50, 0.02, "● Defensa", color="#60a5fa", fontsize=10, ha="center")
fig.text(0.75, 0.02, "● Distribución", color="#4ade80", fontsize=10, ha="center")

plt.tight_layout()
plt.show()
````

### 5️⃣ Heatmap de presencia en el campo

Muestra las zonas del campo donde un jugador tiene más actividad:

````python
# ── HEATMAP DE PRESENCIA ─────────────────────────────────────────────────
from mplsoccer import Pitch
from scipy.stats import gaussian_kde

# Datos de ejemplo: posiciones (x, y) donde el jugador toca el balón
np.random.seed(7)
n_touches = 120

# Simular un mediapunta: más presencia en zona 50-80, centrado
touches_x = np.random.normal(65, 18, n_touches)
touches_y = np.random.normal(40, 15, n_touches)

# Filtrar dentro del campo
touches_x = np.clip(touches_x, 0, 120)
touches_y = np.clip(touches_y, 0, 80)

fig, ax = plt.subplots(figsize=(14, 9))
fig.patch.set_facecolor(DARK_BG)

pitch = Pitch(pitch_type='statsbomb', pitch_color=DARK_BG,
              line_color='#3a3a44', linewidth=1.5)
pitch.draw(ax=ax)

# Heatmap con KDE
pitch.kdeplot(
    touches_x, touches_y,
    ax=ax,
    cmap='Reds',        # Paleta roja → colores Zamora CF
    alpha=0.75,
    levels=15,
    fill=True,
    zorder=2,
    bw_adjust=0.7,
)

# Marcar los toques individuales
pitch.scatter(touches_x, touches_y, ax=ax,
              s=20, color=ZAMORA_WHITE, alpha=0.3, zorder=3)

ax.set_title(f"Heatmap de presencia – Jugador Ejemplo (Mediapunta)\n"
             f"{n_touches} acciones registradas",
             color=ZAMORA_WHITE, fontsize=13, fontweight='bold', pad=15)

plt.tight_layout()
plt.show()
````

## 🗺️ Mapa de Pases

Visualiza la red de pases de un jugador: origen, destino y frecuencia:

````python
# ── MAPA DE PASES ────────────────────────────────────────────────────────
np.random.seed(21)
n_passes = 60

# Simular pases: (x_inicio, y_inicio, x_fin, y_fin)
# Jugador en zona central, pases hacia adelante y laterales
px_start = np.random.normal(55, 10, n_passes)
py_start = np.random.normal(40, 8,  n_passes)
px_end   = px_start + np.random.normal(12, 15, n_passes)
py_end   = py_start + np.random.normal(0,  18, n_passes)

# Clip dentro del campo
px_start = np.clip(px_start, 0, 120)
py_start = np.clip(py_start, 0, 80)
px_end   = np.clip(px_end,   0, 120)
py_end   = np.clip(py_end,   0, 80)

# Clasificar pases: progresivos (avance > 5m) vs laterales/atrás
progresivos = (px_end - px_start) > 5

fig, ax = plt.subplots(figsize=(14, 9))
fig.patch.set_facecolor(DARK_BG)

pitch = Pitch(pitch_type='statsbomb', pitch_color=CARD_BG,
              line_color='#3a3a44', linewidth=1.5)
pitch.draw(ax=ax)

# Pases laterales / hacia atrás
pitch.arrows(px_start[~progresivos], py_start[~progresivos],
             px_end[~progresivos],   py_end[~progresivos],
             ax=ax, width=1.5, headwidth=4, headlength=4,
             color='#3a3a44', alpha=0.5, zorder=3)

# Pases progresivos
pitch.arrows(px_start[progresivos], py_start[progresivos],
             px_end[progresivos],   py_end[progresivos],
             ax=ax, width=2, headwidth=5, headlength=5,
             color=ZAMORA_RED, alpha=0.85, zorder=4)

# Posición media del jugador
ax.scatter(px_start.mean(), py_start.mean(),
           s=400, color=ZAMORA_WHITE, edgecolors=ZAMORA_RED,
           linewidth=3, zorder=6)
ax.text(px_start.mean(), py_start.mean() - 6,
        "Jugador", color=ZAMORA_WHITE, fontsize=9,
        ha='center', fontweight='bold')

# Estadísticas
prog_pct = progresivos.sum() / n_passes * 100
ax.set_title(f"Mapa de Pases – Jugador Ejemplo (Mediocentro)\n"
             f"{n_passes} pases totales · {progresivos.sum()} progresivos ({prog_pct:.0f}%)",
             color=ZAMORA_WHITE, fontsize=13, fontweight='bold', pad=15)

# Leyenda
legend_elements = [
    mpatches.Patch(color=ZAMORA_RED,  label=f'Progresivos ({progresivos.sum()})'),
    mpatches.Patch(color='#3a3a44',   label=f'Laterales/atrás ({(~progresivos).sum()})'),
]
ax.legend(handles=legend_elements, loc='lower left',
          facecolor=CARD_BG, edgecolor='#3a3a44', labelcolor=ZAMORA_WHITE, fontsize=10)

plt.tight_layout()
plt.show()
````

### 🕸️ Red de pases entre jugadores del equipo

Visualiza quién pasa a quién con el grosor proporcional a la frecuencia:

````python
# ── RED DE PASES DEL EQUIPO ───────────────────────────────────────────────
# Posiciones medias de los jugadores en el campo (x, y) StatsBomb
jugadores_equipo = {
    "Portero":      (5,  40),
    "Lat. Dcho":    (28, 68), "Lat. Izdo":    (28, 12),
    "Central D":    (22, 55), "Central I":    (22, 25),
    "M. Defensivo": (42, 40),
    "M. Dcho":      (60, 62), "M. Izdo":      (60, 18),
    "Mediapunta":   (72, 40),
    "Extremo D":    (88, 68), "Extremo I":    (88, 12),
}

# Matriz de pases entre jugadores (simulada)
np.random.seed(5)
nombres = list(jugadores_equipo.keys())
n = len(nombres)
matriz_pases = np.random.randint(2, 25, (n, n))
np.fill_diagonal(matriz_pases, 0)
# Más pases entre vecinos cercanos
for i in range(n):
    for j in range(n):
        xi, yi = jugadores_equipo[nombres[i]]
        xj, yj = jugadores_equipo[nombres[j]]
        dist = np.sqrt((xi-xj)**2 + (yi-yj)**2)
        if dist > 40:
            matriz_pases[i,j] = max(0, matriz_pases[i,j] - 15)

fig, ax = plt.subplots(figsize=(14, 9))
fig.patch.set_facecolor(DARK_BG)

pitch = Pitch(pitch_type='statsbomb', pitch_color=CARD_BG,
              line_color='#3a3a44', linewidth=1.5)
pitch.draw(ax=ax)

max_pases = matriz_pases.max()

# Dibujar líneas de pases
for i, n1 in enumerate(nombres):
    for j, n2 in enumerate(nombres):
        if i >= j: continue
        total = matriz_pases[i,j] + matriz_pases[j,i]
        if total < 8: continue
        x1, y1 = jugadores_equipo[n1]
        x2, y2 = jugadores_equipo[n2]
        alpha  = 0.2 + 0.7 * (total / (max_pases * 2))
        lw     = 1 + 5 * (total / (max_pases * 2))
        ax.plot([x1, x2], [y1, y2],
                color=ZAMORA_RED, alpha=alpha, linewidth=lw, zorder=3)

# Dibujar jugadores
for nombre, (x, y) in jugadores_equipo.items():
    total_pases = matriz_pases[nombres.index(nombre)].sum()
    size = 200 + (total_pases / matriz_pases.sum(axis=1).max()) * 800
    ax.scatter(x, y, s=size, color=ZAMORA_RED, edgecolors=ZAMORA_WHITE,
               linewidth=2, zorder=5)
    offset = 5 if y > 40 else -5
    ax.text(x, y + offset, nombre, color=ZAMORA_WHITE, fontsize=7.5,
            ha='center', va='center', fontweight='bold', zorder=6)

ax.set_title("Red de Pases – Zamora CF\nGrosor = frecuencia · Tamaño = pases totales",
             color=ZAMORA_WHITE, fontsize=13, fontweight='bold', pad=15)

plt.tight_layout()
plt.show()
````
